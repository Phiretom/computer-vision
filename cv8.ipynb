```python
import torch
from diffusers import StableDiffusionPipeline

model_path = "/content/drive/MyDrive/model.ckpt"

pipe = StableDiffusionPipeline.from_single_file(
    model_path,
    torch_dtype=torch.float16,
    load_safety_checker=False,
    low_cpu_mem_usage=False,
    device_map=None
)
pipe = pipe.to("cuda")
pipe.enable_attention_slicing()
```

```python
prompts = {
    #"1": f"ohwx man in cyberpunk city high quality realism ultra realistic 4k",
    #"2": f"ohwx man made of liquid metal, chrome",
    #"3": f"ohwx man in space with visible face, realism",
    #"4": f"ohwx man in fantasy world, high quality",
    #"5": f"ohwx man in fairy tale, high quality",
    #"6": f"ohwx man painting style, realism",
    #"7": f"ohwx man in a forest, high quality, realism",
    #"8": f"ohwx man in a city, high quality, realism",
    #"9": f"ohwx man on a beach, high quality, realism"
}

for name, prompt in prompts.items():
    print(f"Генерация {name}")
    image = pipe(prompt, num_inference_steps=40, guidance_scale=7.5).images[0]

    filename = f"results/result_{name}.png"
    image.save(filename)
```
